# Wear emissions

Tyre, brake and road wear return grams, not g/km, and weight the fleet composition.

```python
from gtfs2emis_py.ef import emi_europe_emep_wear

wear = emi_europe_emep_wear(
    [1.0, 2.0],
    [30.0, 60.0],
    ["TSP", "PM10"],
    ["Ubus Std 15 - 18 t", "Ubus Artic >18 t"],
    [0.5, 0.5],
    process=("tyre", "brake", "road"),
)
wear.pivot_table(index="process", columns="pollutant", values="emi_g", aggfunc="sum")
```
