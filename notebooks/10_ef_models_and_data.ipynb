# EF tables and the other factor models

The EF package exposes one tidy table per model. EMEP is packaged in the wheel;
CETESB, EMFAC and MOVES are fetch-only. This tutorial exercises
the packaged tables and the MOVES/Brazil models without a network call.


```python
from gtfs2emis_py.ef import ef_brazil_cetesb, ef_usa_moves, list_ef_tables, load_ef_table

print(list_ef_tables())
emep = load_ef_table("europe_emep")
print("EMEP rows:", len(emep), "columns:", len(emep.columns))
emep[["Pol", "Segment", "Fuel", "Euro"]].head()
```

```python
cetesb = ef_brazil_cetesb(["CO"], "BUS_URBAN_D", 2015)
moves = ef_usa_moves(["CO"], 2015, reference_year=2019, speed=[10.0, 50.0])
print(cetesb.table[["pollutant", "veh_type", "ef", "unit"]])
moves.table[["pollutant", "speed_kmh", "ef", "unit"]]
```

`load_ef_table(..., fetch=True)` downloads the pinned original `.rda` and
verifies both the raw SHA-256 and the value-level `content_sha256`; a changed
data edition is rejected rather than silently accepted.


```python
from gtfs2emis_py.ef import ef_brazil_scaled_euro

scaled = ef_brazil_scaled_euro(["CO"], 50.0, "BUS_URBAN_D", 2015, "Ubus Std 15 - 18 t", "V")
scaled.table[["pollutant", "veh_type", "euro", "ef", "unit"]]
```
