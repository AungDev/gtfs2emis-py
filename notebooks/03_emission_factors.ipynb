# Emission factors

EMEP factors vary with speed; EMFAC/MOVES use right-closed speed bins.

```python
import numpy as np

from gtfs2emis_py.ef import ef_europe_emep

speeds = np.arange(10, 61, 10)
emep = ef_europe_emep(speeds, "Ubus Std 15 - 18 t", "V", ["CO", "NOx"], fuel="D", tech="SCR")
emep.table[["pollutant", "speed_kmh", "ef", "unit"]]
```

```python
from gtfs2emis_py.ef import ef_usa_emfac

emfac = ef_usa_emfac(["CO"], 2020, "D", model_year=2015, speed=[8.0, 8.1, 16.0, 16.1])
emfac.table[["speed_kmh", "bin_lower_kmh", "bin_upper_kmh", "ef"]]
```
