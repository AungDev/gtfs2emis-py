# Units and pollutants

The package computes on canonical floats: km, km/h, g/km, g, m and seconds.
The pollutant registry distinguishes a factor unit from the emitted value unit.


```python
from gtfs2emis_py.units import UnitError, as_g, as_g_per_km, as_km, as_kmh

print(as_km(1000, "m"), as_kmh(10, "m/s"), as_g_per_km(2, "kg/km"), as_g(1.5, "kg"))
try:
    as_km(-1, "km")
except UnitError as error:
    print(type(error).__name__, error)
```

```python
from gtfs2emis_py.pollutants import emission_value_unit, list_pollutants, pollutant_unit

print("pollutants:", len(list_pollutants()))
print("EC factor unit (MOVES):", pollutant_unit("EC", model="ef_usa_moves"))
print("EC value unit (MOVES):", emission_value_unit("EC", model="ef_usa_moves"))
```

The factor unit (`g/km`, `MJ/km`, `kJ/km`, `#/kWh`) is separate from the
emitted value unit (`g`, `MJ`, `kJ`, `#`); `EmissionResult` keeps both.
