# Getting started

Build a transport table from the bundled Curitiba feed, load its fleet and estimate hot-exhaust emissions.

```python
import contextlib
import io
from pathlib import Path

from gtfs2emis_py import emission_model, load_fleet, transport_model

DATA = Path.cwd() / "tests/data"
with contextlib.redirect_stdout(io.StringIO()):
    tp = transport_model(DATA / "bra_cur_gtfs.zip")
print(f"{len(tp.table):,} segments across {tp.table['shape_id'].nunique()} shapes")
tp.table[["shape_id", "speed_kmh", "dist_km"]].head()
```

```python
fleet = load_fleet(DATA / "bra_cur_fleet.txt", ef_model="ef_brazil_cetesb")
result = emission_model(tp, "ef_brazil_cetesb", fleet, ["CO", "NOx"])
result.emi.head()
```

```python
from gtfs2emis_py import emis_summary

emis_summary(result)
```
