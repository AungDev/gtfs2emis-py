# Reading and validating a GTFS feed

`read_gtfs()` wraps `gtfs-kit` with structural validation, ID normalisation,
distance-unit normalisation and an explicit cleaning report. The committed
synthetic feed keeps this tutorial offline.


```python
from pathlib import Path

from gtfs2emis_py.gtfs import read_gtfs

DATA = Path.cwd() / "tests/data"
feed = read_gtfs(DATA / "synthetic_edge_cases.zip")
print("valid:", feed.ok)
print("distance units:", feed.dist_units)
print("stops:", len(feed.feed.stops), "trips:", len(feed.feed.trips))
```

```python
# clean=True records the rows it removed; validation warnings are separate
# from hard validation errors.
print(feed.clean_report)
print(feed.validation.warnings[:3])
```

```python
# validate=False leaves ok=None, not True: an unchecked feed cannot be
# mistaken for a validated one.
unchecked = read_gtfs(DATA / "synthetic_edge_cases.zip", validate=False)
print("unchecked ok:", unchecked.ok)
```
